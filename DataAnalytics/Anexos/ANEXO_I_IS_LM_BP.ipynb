```python
import matplotlib.pyplot as plt
import numpy as np

# Criar gráfico IS-LM-BP estilizado
fig, ax = plt.subplots(figsize=(8,6))

# Definir espaço
Y = np.linspace(0, 100, 200)
i = np.linspace(0, 10, 200)

# Curvas simples
IS = 5 - 0.03*Y  # decrescente
LM = 2 + 0.02*Y  # crescente
BP = 4*np.ones_like(Y)  # linha horizontal

ax.plot(Y, IS, label="IS (Demanda agregada)", color="blue")
ax.plot(Y, LM, label="LM (Mercado monetário)", color="red")
ax.plot(Y, BP, label="BP (Balanço de Pagamentos)", color="green")

# Anotações deslocamentos
ax.annotate("Choque externo\n(IS desloca à esquerda)", xy=(40, 3.8), xytext=(15, 6),
            arrowprops=dict(arrowstyle="->", color="blue"), fontsize=9, color="blue")
ax.annotate("Fuga de capitais\n(LM para cima/esquerda)", xy=(70, 3.5), xytext=(80, 7),
            arrowprops=dict(arrowstyle="->", color="red"), fontsize=9, color="red")

# Configurações visuais
ax.set_title("Modelo IS-LM-BP em Economia Aberta", fontsize=14)
ax.set_xlabel("Renda (Y)", fontsize=12)
ax.set_ylabel("Taxa de Juros (i)", fontsize=12)
ax.legend()
ax.grid(True, linestyle="--", alpha=0.6)

plt.tight_layout()
plt.show()   # <-- mostra direto no notebook
```
